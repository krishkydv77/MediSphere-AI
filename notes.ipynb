### **2. Project Workflow (Karam-dar-Karam Kaise Kaam Karta Hai?)**

Aapka poora system 4 mukhya layers me kaam karta hai:

1. **Client & UI Layer (`templates/`):**
* **`index.html`** par user chat ya injury image upload karta hai.
* **`admin.html`** par admin login karke system credentials generate karta hai.
* **`system.html`** par system monitor servers ka health check dekhta hai.


2. **API & Security Layer (`main_2.py` & `user_handle.py`):**
* FastAPI requests ko receive karta hai. **SlowAPI** rate-limiting check karta hai (DDoS/Spam protection ke liye).
* **JWT / OAuth (Google)** security verify karta hai ki request valid user ya admin ki hai ya nahi.


3. **AI & MCP Agent Layer (`medical_mcp/` & `chatbot/`):**
* Jab user kuch puchta hai (jaise *"mere lungs me problem hai"*), toh `llm_helper.py` aur `ai_agent.py` activate hote hain.
* Agent **FastMCP** tools ke zariye MySQL database se query karke relevant specialist (**Pulmonologist**) aur available doctors ki list fetch karta hai.


4. **Database & Booking Layer (`chatbot/database/`):**
* User doctor aur slot select karta hai, jiske baad SQLAlchemy securely booking ko MySQL database me save kar deti hai (`Booking` table).
